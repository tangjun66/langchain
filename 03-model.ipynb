```python
from langchain.chat_models import  init_chat_model
#从.env文件中加载环境变量
from dotenv import load_dotenv # pip install python-dotenv

#加载环境变量
load_dotenv() #默认会自动加载当前目录下的.env文件，并将文件中的内容加载到环境变量中
```

```python
import os
os.getenv("BASE_URL")
```

```python
# 模型定义
model = init_chat_model(
    model=os.getenv("BASE_MODEL"),
    temperature=0.7,
    max_tokens=2048,
    timeout=30,
    api_key=os.getenv("BASE_API_KEY"),
    base_url=os.getenv("BASE_URL"),
)
```

```python
# 模型调用
response = model.invoke("你好，请介绍一下自己！")
```

```python
response
```

```python
# 定义字典格式
message = [
    {"role": "system", "content": "你是一个翻译专家，擅长将中文翻译成英文。"}, #SystemMessage
    {"role": "user", "content": "请将以下中文翻译成英文：今天天气真好！"} #HumanMessage
]
response = model.invoke(message)
```

```python
response
```

```python
response.content
```

```python
from langchain_core.messages import SystemMessage, HumanMessage
# 定义具体消息类型
message = [
    SystemMessage(content="你是一个翻译专家，擅长将中文翻译成英文。"),
    HumanMessage(content="请将以下中文翻译成英文：今天天气真好！")
]
response = model.invoke(message) #同步过程
print(response)
```

```python
response
```

```python
message = [
    SystemMessage(content="你是一个人工智能专家，擅长解释人工智能的概念。"),
    HumanMessage(content="请详细解释一下什么是人工智能？")
]
for chunk in model.stream(message): #model.astream(message) #异步流式
    print(chunk.text, end="", flush=True)
```

```python
 # ajax|fetch
 # 拓展：异步编程：多线程
response = model.ainvoke(message) #异步过程
print(response)#协程对象，就是异步对象，而不是异步执行的结果
```

```python
 # ajax|fetch
 # 拓展：异步编程：多线程
response = await model.ainvoke(message) #异步过程, 等待异步执行的结果
print(response)#协程对象，就是异步对象，而不是异步执行的结果
```

```python
#批量回复
questions = ["今天天气怎么样？", "你喜欢什么颜色？", "你会什么语言？"]
response = model.batch(questions) # model.abatch(questions) #异步批量
```

```python
for r in response:
    print(r.content)
```
